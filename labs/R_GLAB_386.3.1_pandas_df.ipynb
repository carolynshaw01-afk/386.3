```python
import numpy as np
import pandas as pd
```

```python
# Example #1: Dictionary d with lists [1, 2, 3] and [2, 4, 8]
# Dictionary to make into a data frame
d = {
    'x': [1, 2, 3],
    'y': [2, 4, 8],
    'z': 100 # scalar values, repeated for all records
}

print(pd.DataFrame(d))
```

```python
# Example #2: 
# Use the index argument to create custom index values (index parameter)
# Use columns argument/parameter to select columns or give names if none present
print(pd.DataFrame(d, index=[100, 200, 300], columns=['x', 'y', 'a']))
```

```python
# Example #3:  The student_dict with Names, Ages, and Marks
student_dict = {
    'Name': ['Joe', 'Nat', 'Harry'],
    'Age': [20, 21, 28],
    'Marks': [85.10, 77.80, 91.54]
}

print(pd.DataFrame(student_dict))
```

```python
# Example #4: List of Dictionaries vs. List of structural formats

# Format A: List of Dictionaries
data_dict_list = [
    {'x': 1, 'y': 2, 'z': 3},
    {'x': 2, 'y': 10, 'z': 15},
    {'x': 4, 'y': 13, 'z': 16}
]
print(pd.DataFrame(data_dict_list))
```

```python
# Example #5:  The Patient ID, Name, and DOB horizaontal zip() combination table
# Separate tracking lists for raw patient records
patientID = [101, 102, 103, 104]
name = ['Daniel', 'John', 'Abbi', 'Charlie']
dob = ['2023-01-01', '2024-01-02', '3/10/1987 14045', '13th of October, 2026']

# zip() combines multiple lists horizontally into iterable row rows
pt_df = pd.DataFrame(zip(patientID, name, dob))

# zip() - function combines lists to make a tuple of tuples
# tuples are an iterator, which is what the DF function takes
pt_df = pd.DataFrame(zip(patientID, name, dob))

print(pt_df)
```

```python
# Example #6:  The Stock Ticker and Price tracking metadata table
#  Stock tracking lists
stocks = ['IBM', 'APPLE', 'TWTTER', 'GE', 'MSFT']
prices = [115.00, 119.14, 19.77, 25.99, 26]

# Convert lists into a stock tracking DataFrame
stocks_df = pd.DataFrame(zip(stocks, prices), columns=['Ticker', 'Price'])

print(stocks_df)
```
